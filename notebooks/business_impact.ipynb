```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from src.data import X_train, y_train, X_test, y_test

model = LogisticRegression()
model.fit(X_train, y_train)

y_score = model.predict_proba(X_test)[:,1]
y = np.asarray(y_test)

top_k_idx = np.argsort(y_score)[::-1][:100]
```

```python
x= X_test.reset_index(drop=True).loc[top_k_idx]
y=y[top_k_idx]
```

```python
x['Churn'] = y
x['Cost of outflow'] = 0
less_then_3_tenure_mask = x['Tenure']<3
x['Cost of outflow'].loc[less_then_3_tenure_mask] = (x.loc[less_then_3_tenure_mask]['Salary']/12+x.loc[less_then_3_tenure_mask]['Salary']/12/x.loc[less_then_3_tenure_mask]['Average Monthly Hours Worked']*x.loc[less_then_3_tenure_mask]['Training Hours'])*x.loc[less_then_3_tenure_mask, 'Churn']
less_then_9_tenure_mask = x['Tenure']<9
x['Cost of outflow'].loc[less_then_9_tenure_mask] = (x.loc[less_then_9_tenure_mask]['Salary']/12*2+x.loc[less_then_9_tenure_mask]['Salary']/12/x.loc[less_then_9_tenure_mask]['Average Monthly Hours Worked']*x.loc[less_then_9_tenure_mask]['Training Hours'])*x.loc[less_then_9_tenure_mask, 'Churn']
greater_then_8_tenure_mask = x['Tenure']>8
x['Cost of outflow'].loc[greater_then_8_tenure_mask] = (x.loc[greater_then_8_tenure_mask]['Salary']/12*3+x.loc[greater_then_8_tenure_mask]['Salary']/12/x.loc[greater_then_8_tenure_mask]['Average Monthly Hours Worked']*x.loc[greater_then_8_tenure_mask]['Training Hours'])*x.loc[greater_then_8_tenure_mask, 'Churn']
```

```python
x['Cost of intervention'] = 5000
x['score'] = x['Cost of outflow'] - x['Cost of intervention']
x['score'].sum()
```
